```python
import numpy as np
import cv2

# 11x11 matris, ortada parlak bölge
img = np.zeros((11, 11), dtype=np.uint8)
img[3:8, 3:8] = 200

# Tuz-biber gürültüsü (birkaç nokta)
img_noisy = img.copy()
img_noisy[4, 4] = 255   # tuz
img_noisy[6, 6] = 0     # biber
img_noisy[5, 3] = 255   # tuz
img_noisy[3, 7] = 0     # biber

print("Gürültülü:\n", img_noisy)
print()
print("Blur:\n", cv2.blur(img_noisy, (3,3)))
print()
print("Gaussian:\n", cv2.GaussianBlur(img_noisy, (3,3), 0))
print()
print("Median:\n", cv2.medianBlur(img_noisy, 3))
print()
print("Bilateral:\n", cv2.bilateralFilter(img_noisy, 5, 75, 75))
```
