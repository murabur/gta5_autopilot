```python
import cv2
import numpy as np
import matplotlib.pyplot as plt #görselleştirme için matplotlib kütüphanesi
```

```python
img = np.zeros((512, 512, 3), np.uint8) #np.zeros ile 0'lardan oluşan 512 - 512 bir matris tanımlıyoruz.

# Çizgi: (Resim, Başlangıç, Bitiş, Renk(BGR), Kalınlık)
cv2.line(img, (0, 0), (512, 512), (0, 255, 0), 3)

#yorum satırlarını kaldırıp denemek size kalmış. Hücreyi her çalıştırdığınızda önce img= np.zeros ile matris sıfırlanıyor bunu unutmayın.
#cv2.line(img, (0, 0), (512, 100), (0, 255, 0), 3)
#cv2.line(img, (0, 0), (512, 300), (0, 255, 0), 3)
#cv2.line(img, (0, 0), (512, 800), (0, 255, 0), 3)  #512 sınırı aşıldığı için çizgi y ekseninde 800'e gitmeye çalışıyor. Y ekseni dikeyde yukarıdan aşağı doğru artıyor.


plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB)) # Notebook'ta doğru renk için RGB'ye çeviriyoruz
plt.show()

#not: "OpenCV'de koordinat sistemi sol üstten başlar. X ekseni soldan sağa, Y ekseni yukarıdan aşağıya artar. Çizim yaparken (x, y) formatı kullanılır."
```

```python
img = np.zeros((512, 512, 3), np.uint8) #np.zeros ile 0'lardan oluşan 512 - 512 bir matris tanımlıyoruz.

# görsel - sol_üst_köşe - sağ_alt_köşe - BGR_renk - kalınlık
cv2.rectangle(img=img, pt1=(100, 100), pt2=(400, 400), color=(255, 0, 0), thickness=5)

plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB)) # Notebook'ta doğru renk için RGB'ye çeviriyoruz
plt.show()
```

```python
img = np.zeros((512, 512, 3), np.uint8) #np.zeros ile 0'lardan oluşan 512 - 512 bir matris tanımlıyoruz.


#görsel - merkez - yarıçap - BGR renk - Kalınlık
cv2.circle(img=img, center=(256, 256), radius=50, color=(0, 255, 0), thickness=3)


plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB)) # Notebook'ta doğru renk için RGB'ye çeviriyoruz
plt.show()
```
