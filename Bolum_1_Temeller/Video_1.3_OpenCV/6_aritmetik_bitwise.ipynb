```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# resmi oku
img = cv2.imread("ornek_goruntu.png")

# BGR → RGB (matplotlib için gerekli)
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# parlaklık artır
bright_numpy = img + np.ones(img.shape, dtype=np.uint8) * 50

bright_cv2 = cv2.add(img, np.ones(img.shape, dtype=np.uint8) * 50)

# matplotlib ile göster
plt.figure(figsize=(10,5))

plt.subplot(1,3,1)
plt.imshow(img)
plt.title("Original")
plt.axis("on")

plt.subplot(1,3,2)
plt.imshow(bright_numpy)
plt.title("Numpy")
plt.axis("on")

plt.subplot(1,3,3)
plt.imshow(bright_cv2)
plt.title("CV2")
plt.axis("on")



plt.show()
```

```python
pixel = np.array([250], dtype=np.uint8)
ten = np.array([10], dtype=np.uint8)

result = cv2.add(pixel, ten, dtype=cv2.CV_8U)

#uint8 tipinin üst sınırı 255'dir. Numpy'da 250 + 10 yaptığınızda 4 sonucu alırsınız çünkü 255i geçtiği vakit 0'dan başlar. OpenCV add fonksiyonunda 255e gelindiği vakit ne eklerseniz ekleyin
#255'de kalır. Üstteki kod çalıştırıldığında görüleceği üzere numpy'da beyazlar siyaha kaymışken opencv'de görsel genel olarak daha parlak bir hale geldi.

print(result)
```
