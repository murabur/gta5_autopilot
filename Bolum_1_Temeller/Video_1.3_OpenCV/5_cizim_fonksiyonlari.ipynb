```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = np.zeros((512, 512, 3), np.uint8) # np.zeros ile 0'lardan oluşan 512x512 siyah bir tuval (matris) tanımlıyoruz.

# cv2.ellipse parametreleri sırasıyla:
# img: çizim yapılacak görsel
# center: elipsin merkez koordinatı (x, y)
# axes: eksen uzunlukları (yatay_yarıçap, dikey_yarıçap) -> Elipsin ne kadar basık/uzun olacağını belirler
# angle: elipsin kendi etrafındaki dönüş açısı (derece cinsinden)
# startAngle: çizimin başlayacağı açı (Tam bir elips için 0)
# endAngle: çizimin biteceği açı (Tam bir elips için 360, yarım elips için 180 yapabilirsiniz)
# color: BGR renk formatı
# thickness: Çizgi kalınlığı (-1 yaparsanız içini tamamen boyar)

cv2.ellipse(img=img, center=(256, 256), axes=(100, 200), angle=90, startAngle=0, endAngle=360, color=(255, 0, 0), thickness=3)

plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB)) # Notebook'ta doğru renk (Mavi) için BGR'den RGB'ye çeviriyoruz
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# 512x512 siyah bir tuval oluşturuyoruz
img = np.zeros((512, 512, 3), np.uint8)

# --- ÇOKGEN NOKTALARINI TANIMLAMA ---
# Önemli: Noktalar [x, y] şeklinde bir liste olmalı.
# Burada bir üçgenin köşelerini belirliyoruz:
pts = np.array([[256, 100], [100, 400], [412, 400]], np.int32)

# OpenCV, noktaları [Nokta_Sayısı, 1, 2] boyutunda bir matris olarak bekler.
# .reshape((-1, 1, 2)) işlemi bu uyumu sağlar.
#-1 = eleman sayısını otomatik say
# 1 = grup içindeki sayı, her elemanın tek bir grubu olsun
# 2 = koordinat çifti - en içteki her pakette 2 tane sayı (x,y) olsun.
#opencv için bu formatta verilmesi gerekli
pts = pts.reshape((-1, 1, 2)) #-1 

#Reshape Öncesi (3x2 matris): [[x1, y1], [x2, y2], [x3, y3]]
#Reshape Sonrası (3x1x2 matris): [[[x1, y1]], [[x2, y2]], [[x3, y3]]]

# cv2.polylines parametreleri:
# img: Çizim yapılacak görsel
# pts: Noktalar dizisi (Liste içinde gönderilmelidir: [pts])
# isClosed: True ise son noktayı ilk noktaya bağlar (şekli kapatır), False ise açık bırakır.
# color: BGR renk formatı (Burada Yeşil: 0, 255, 0)
# thickness: Çizgi kalınlığı

cv2.polylines(img=img, pts=[pts], isClosed=True, color=(0, 255, 0), thickness=5)

plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.show()
```

Soru: Neden pts = pts değil de pts = [pts]?
Cevap: Çünkü OpenCV bu fonksiyonda bir "eşya" değil, bir "eşya listesi" bekler. Tek bir üçgen çizecek olsak bile, o üçgeni bir listenin içine koyarak (konteynır gibi) OpenCV'ye teslim ederiz.


```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# 512x512 siyah bir tuval oluşturuyoruz
img = np.zeros((512, 512, 3), np.uint8)

# --- ÇOKGEN NOKTALARINI TANIMLAMA ---
# Yine bir üçgenin köşelerini belirliyoruz:
pts = np.array([[256, 100], [100, 400], [412, 400]], np.int32)

# ÖNEMLİ: reshape mantığı polylines ile TAMAMEN AYNI!
# Veriyi OpenCV'nin istediği "özel paketleme" formatına sokuyoruz.
pts = pts.reshape((-1, 1, 2))


# cv2.fillPoly parametreleri:
# img: Çizim yapılacak görsel
# pts: Noktalar dizisi (Yine liste içinde [pts] şeklinde verilmelidir)
# color: BGR renk formatı (Burada Turkuaz/Cam Göbeği: 255, 255, 0)
# NOT: thickness parametresi YOKTUR. Çünkü bu fonksiyon alanı tamamen doldurur.

cv2.fillPoly(img=img, pts=[pts], color=(255, 255, 0))

# Notebook'ta doğru renk gösterimi için RGB'ye çeviriyoruz
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# 512x512 siyah bir tuval oluşturuyoruz
img = np.zeros((512, 512, 3), np.uint8)

# --- 1. ŞEKİL: ÜÇGEN ---
# Koordinatlar: Üst merkez, sol alt, sağ alt
pts1 = np.array([[100, 100], [50, 250], [150, 250]], np.int32).reshape((-1, 1, 2))

# --- 2. ŞEKİL: KARE ---
# Koordinatlar: Sol üst, sağ üst, sağ alt, sol alt
pts2 = np.array([[300, 100], [450, 100], [450, 250], [300, 250]], np.int32).reshape((-1, 1, 2))

# --- ÇOKLU ÇİZİM ---
# pts=[pts1, pts2] yazarak OpenCV'ye diyoruz ki:
# "Sana bir liste veriyorum, bu listenin içindeki tüm paketleri boya!"
cv2.fillPoly(img=img, pts=[pts1, pts2], color=(0, 255, 255)) # Sarımsı/Turkuaz renk

# Sonucu göster
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# 512x512 siyah bir tuval oluşturuyoruz
img = np.zeros((512, 512, 3), np.uint8)

# --- METİN YAZDIRMA (FONKSİYON İÇİNDE TANIMLAMA) ---

cv2.putText(
    img=img,                          # Yazının çizileceği görsel 
    text="HIZ: 80 KM/H",              # Ekranda görünecek metin (Mutlaka String olmalı)
    org=(50, 250),                    # Metnin sol alt köşesinin başlangıç koordinatı (x, y)
    fontFace=cv2.FONT_HERSHEY_SIMPLEX,# Yazı tipi kütüphanesi 
    fontScale=1.5,                    # Yazının büyüklük katsayısı (1 = standart boyut)
    color=(0, 255, 0),                # Yazı rengi (BGR formatında Yeşil)
    thickness=3,                      # Harflerin çizgi kalınlığı
    lineType=cv2.LINE_AA              # Kenar yumuşatma (Yazının tırtıklı görünmesini engeller)
)

# Notebook üzerinde doğru renkleri görmek için BGR -> RGB çevrimi yapıyoruz
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# 512x512 siyah bir tuval oluşturuyoruz
img = np.zeros((512, 512, 3), np.uint8)

# 1. ADIM: Yazının boyutlarını ölçüyoruz
(width, height), baseline = cv2.getTextSize(
    text="HIZ: 80 KM/H",
    fontFace=cv2.FONT_HERSHEY_SIMPLEX,
    fontScale=1.5,
    thickness=2
)

# --- YENİ: PADDING (İÇ BOŞLUK) DEĞERİ ---
p = 10 # Her yönden 10 piksel boşluk bırakacağız

# 2. ADIM: Arka plan kutusunu (Dikdörtgeni) çiz
cv2.rectangle(
    img=img,
    pt1=(50, 150), # Başlangıç noktası
    # DÜZELTME: pt2 hesaplanırken hem genişliğe hem yüksekliğe (2 * p) ekliyoruz.
    # Böylece yazı içeride kalıyor, kutu dışarı doğru esniyor.
    pt2=(50 + width + (2 * p), 150 + height + baseline + (2 * p)), 
    color=(0, 255, 0), # Yeşil
    thickness=-1
)

# 3. ADIM: Yazıyı kutunun içine bas
cv2.putText(
    img=img,
    text="HIZ: 80 KM/H",
    # DÜZELTME: org noktasını (50 + p, 150 + height + p) yaparak 
    # yazıyı kutunun sol üst köşesinden çaprazlama içeri itiyoruz.
    org=(50 + p, 150 + height + p), 
    fontFace=cv2.FONT_HERSHEY_SIMPLEX,
    fontScale=1.5,
    color=(0, 0, 0), # Siyah
    thickness=2,
    lineType=cv2.LINE_AA
)

plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# 512x512 siyah bir tuval oluşturuyoruz
img = np.zeros((512, 512, 3), np.uint8)

# --- OK ÇİZİMİ ---

cv2.arrowedLine(
    img=img,              # Okun çizileceği görsel 
    pt1=(256, 450),       # Okun başlangıç noktası 
    pt2=(256, 150),       # Okun bittiği (ucu) nokta 
    color=(0, 255, 255),  # Okun rengi (BGR formatında Sarı)
    thickness=5,          # Okun gövde ve uç kalınlığı
    tipLength=0.2         # Ok ucunun toplam uzunluğa oranı (0.1 ile 0.5 arası idealdir)
)

# Notebook üzerinde doğru renkleri görmek için BGR -> RGB çevrimi yapıyoruz
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# 512x512 siyah bir tuval oluşturuyoruz
img = np.zeros((512, 512, 3), np.uint8)

# --- MARKER (İŞARETÇİ) ÇİZİMİ ---

cv2.drawMarker(
    img=img,                      # İşaretin çizileceği görsel
    position=(256, 256),          # İşaretin tam merkez koordinatı (x, y)
    color=(0, 0, 255),            # İşaretin rengi (BGR formatında Kırmızı)
    markerType=cv2.MARKER_CROSS,  # İşaret tipi (Artı, Çarpı, Kare, Elmas vb.)
    markerSize=30,                # İşaretin toplam genişliği/boyutu (piksel)
    thickness=3                   # İşareti oluşturan çizgilerin kalınlığı
)

# Notebook üzerinde doğru renkleri görmek için BGR -> RGB çevrimi yapıyoruz
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.show()
```
