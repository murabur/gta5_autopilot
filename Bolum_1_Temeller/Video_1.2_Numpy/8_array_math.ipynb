NumPy’de matematik işlemleri element-wise çalışır.
Yani işlem array’in tamamına uygulanır, tek tek döngü yazılmasına gerek yoktur.


```python
import time

t0 = time.time()
liste= []
for i in range(1_000_000):
    us = i ** 100
    liste.append(us)
t1 = time.time()

print(t1 - t0)

```

```python
import time
import numpy as np


t0 = time.time()

numpy_liste = np.arange(1_000_000)**100

t1 = time.time()

print(t1-t0)
```
