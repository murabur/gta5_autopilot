Numpy random modülü rastgele sayı üretmek için kullanılır.

Rastgelelik gerçek rastgelelik değildir ama çoğu uygulama için yeterlidir.

```python
#np.random.seed() tekrarlanabilir rastgelelik için kullanılır.

import numpy as np

np.random.seed(42)
print(np.random.rand(5))

np.random.seed(42)
print(np.random.rand(5))
```

```python
#0 ile 1 arasında uniform dağılımlı rastgele sayı üretir.

import numpy as np

print(np.random.rand())
```

```python
import numpy as np

print(np.random.rand(5)) #1D array
```

```python
import numpy as np

print(np.random.rand(2,3)) #2D Array
```

```python
#normal dağılımda rastgele saytı üretir.
#pozitif veya negatif değer olabilir.
#Değerler çoğunlukla 0 çevresindedir.

import numpy as np

print(np.random.randn(5))
```

```python
#belli aralıkta rastele tam sayı üretir.

import numpy as np

print(np.random.randint(0,10)) #0 dahil 10 hariç
```

```python
import numpy as np

print(np.random.randint(0,256,size=(2,3))) #2D Array
```

```python
#belli aralıkta uniform dağılımlı float üretir.
import numpy as np

print(np.random.uniform(10,20))
```

```python
import numpy as np

print(np.random.uniform(10,20,size=(2,3))) #array
```

```python
#belirli ortalama ve standart sapma ile normal dağılımdan sayı üretir.


import numpy as np

print(np.random.normal(loc=0, scale=1))
```

```python
import numpy as np

print(np.random.normal(loc=100, scale=15, size=5))
```

```python
#bir listeden veya aralıktan rastgele seçim yapar.

import numpy as np

arr = np.array([10,20,30,40,50])

print(np.random.choice(arr))
```

```python
#birden fazla seçim

import numpy as np

arr = np.array([10,20,30,40,50])

print(np.random.choice(arr, size=3))
```

```python
#tekrarsız seçim
import numpy as np

arr = np.array([10,20,30,40,50])

print(np.random.choice(arr, size=3, replace=False))
```

```python
#Bir array'i yerinde karıştırır

import numpy as np

arr = np.array([1,2,3,4,5])

print("önce:", arr)

np.random.shuffle(arr)

print("sonra:", arr)
```

```python
#karıştırılmış yeni bir kopya döndürür

import numpy as np

arr = np.array([1,2,3,4,5])

perm = np.random.permutation(arr)

print("arr:", arr)
print("perm:", perm)
```

Uniform dağılımda belirli aralıktaki tüm değerlerin gelme olasılığı eşittir.

Normal dağılımda, ortalama etrafında yoğunlaşan çan eğrisi dağılımıdır.
