np.where()'in iki türlü kullanımı vardır.
- np.where(koşul) = koordinat/indeks verir
- np.where(koşul, A, B) = koşul geçerliyse A, geçerli değilse B

```python
import numpy as np

arr = np.array([10, 50, 120, 210])
indices = np.where(arr > 100)
print(indices)
```

```python
import numpy as np

arr = np.array([5, 12, 18, 25, 30, 45])

idx = np.where(arr > 20)

print(idx)
print(arr[idx])
```

```python
arr = np.array([
    [10, 220, 30],
    [150, 80, 240],
    [60, 200, 15]
])

coords = np.where(arr > 200)
print(coords) #satır indeksleri - sütun indeksleri
```

```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
import os

current_dir = os.getcwd()

parent_dir = os.path.dirname(current_dir)
parent_dir = os.path.dirname(parent_dir)

img_path = os.path.join(parent_dir, "assets", "ornek_goruntu.png")

print(f"Erişilen Yol: {img_path}")

img = cv2.imread(img_path)

if img is None:
    print("Görüntü okunamadı!")
    exit()

print("Görüntü başarıyla yakalandı")

gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# boolean mask üret
mask = gray > 200

# koordinatları al
ys, xs = np.where(mask)

print(f"Koşulu sağlayan piksel sayısı: {len(xs)}")

# matplotlib ile göster
plt.figure(figsize=(10,5))

# orijinal görüntü
plt.subplot(1,2,1)
plt.imshow(gray, cmap="gray")
plt.title("Grayscale Image")
plt.axis("off")

# mask görüntüsü
plt.subplot(1,2,2)
plt.imshow(mask, cmap="gray")
plt.title("Bright Pixel Mask")
plt.axis("off")

plt.show()
```

```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
import os

current_dir = os.getcwd()

parent_dir = os.path.dirname(current_dir)
parent_dir = os.path.dirname(parent_dir)

img_path = os.path.join(parent_dir, "assets", "ornek_goruntu.png")

print(f"Erişilen Yol: {img_path}")

img = cv2.imread(img_path)

if img is None:
    print("Görüntü okunamadı!")
    exit()

print("Görüntü başarıyla yakalandı")

gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# boolean mask üret
mask = gray > 230

# koordinatları al
ys, xs = np.where(gray > 230)
center_x = int(xs.mean())
center_y = int(ys.mean())

print("center_x: " ,center_x)
print("center_y: ", center_y)

cv2.circle(gray, center = (center_x,center_y), radius= 5, color = 0, thickness= 500) #parlaklık merkezine circle çiz
print(f"Koşulu sağlayan piksel sayısı: {len(xs)}")

# matplotlib ile göster
plt.figure(figsize=(10,5))

# orijinal görüntü
plt.subplot(1,2,1)
plt.imshow(gray, cmap="gray")
plt.title("Grayscale Image")
plt.axis("off")

# mask görüntüsü
plt.subplot(1,2,2)
plt.imshow(mask, cmap="gray")
plt.title("Bright Pixel Mask")
plt.axis("off")

plt.show()
```

```python
import numpy as np
import cv2
import os
import matplotlib.pyplot as plt

current_dir = os.getcwd()
parent_dir = os.path.dirname(current_dir)
parent_dir = os.path.dirname(parent_dir)


img_path = os.path.join(parent_dir, "assets", "ornek_goruntu.png")

print(f"Erişilen yol: {img_path} " )

img = cv2.imread(img_path)
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

mask = gray > 254

ys, xs = np.where(mask)

x_min = xs.min()
x_max = xs.max()
y_min = ys.min()
y_max = ys.max()

print(x_min)
print("Ortalama y:", ys.mean())
output = img.copy()

cv2.rectangle(output, (x_min, y_min), (x_max, y_max), (0,255,0), 5)

# matplotlib ile göster
plt.figure(figsize=(10,5))

# orijinal görüntü
plt.subplot(1,2,1)
plt.imshow(cv2.cvtColor(output, cv2.COLOR_BGR2RGB))
plt.title("Grayscale Image")
plt.axis("off")

# mask görüntüsü
plt.subplot(1,2,2)
plt.imshow(mask, cmap="gray")
plt.title("Bright Pixel Mask")
plt.axis("off")

plt.show()
```
