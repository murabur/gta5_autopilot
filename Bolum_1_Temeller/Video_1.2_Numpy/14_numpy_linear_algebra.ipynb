## Linear algebra nedir

Linear algebra, vektörler ve matrislerle yapılan matematiksel işlemler alanıdır.

NumPy tarafında bu bölümün temel amacı şudur:

- vektör nedir
- matris nedir
- iki vektör nasıl çarpılır
- iki matris nasıl çarpılır
- bir matrisin tersi nedir
- determinant nedir


```python
#vektör
import numpy as np

v = np.array([1, 2, 3])

print("v:", v)
print("v.shape:", v.shape)
```

```python
#matris
import numpy as np

A = np.array([
    [1, 2],
    [3, 4]
])

print("A:\n", A)
print("A.shape:", A.shape)
```

Dot product, iki vektörün çarpılıp toplanmasıdır.
a · b = a1 x b1 + a2 x b2 + a3 x b3 + ...

```python
import numpy as np

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print("a:", a)
print("b:", b)

print("np.dot(a, b):", np.dot(a, b))
print("a @ b:", a @ b)
```

```python
import numpy as np

A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

print("A:\n", A)
print("B:\n", B)

print("A * B:\n", A * B) #element - wise çarpım

print("A @ B:\n", A @ B) #matris çarpımnı

#matris çarpımı için boyut şartı (m x n) @ (n x p)
```

```python
#çarpılamayan örnek

A = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

C = np.array([
    [1, 2],
    [3, 4]
])

print("A.shape:", A.shape)
print("C.shape:", C.shape)
print(A @ C)
```

```python
import numpy as np

A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

print("np.matmul(A, B):\n", np.matmul(A, B)) #@ operatörünün karşılığı np.matmul'dur.
```

```python
#inverse - ters matris
#bir matrisin inverse'ü onu çarptığında birim matris veren tersidir

import numpy as np

A = np.array([
    [1, 2],
    [3, 4]
])

A_inv = np.linalg.inv(A)

print("A:\n", A)
print("A_inv:\n", A_inv)
print("A @ A_inv:\n", A @ A_inv)
```

```python
#determinant

import numpy as np

A = np.array([
    [1, 2],
    [3, 4]
])

print("A:\n", A)
print("det(A):", np.linalg.det(A))
```
