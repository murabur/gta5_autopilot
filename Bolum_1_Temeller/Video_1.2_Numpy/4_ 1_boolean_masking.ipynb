```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
import os

current_dir = os.getcwd()

parent_dir = os.path.dirname(current_dir)
parent_dir = os.path.dirname(parent_dir)

img_path = os.path.join(parent_dir, "assets", "ornek_goruntu.png")

print(f"Erişilen Yol: {img_path}")

img = cv2.imread(img_path)

if img is None:
    print("Görüntü okunamadı!")
    exit()

print("Görüntü başarıyla yakalandı")

# grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY) #DİKKAT - GÖRÜNTÜ GRAY SCALE OKUNUYOR

# boolean mask
mask = gray > 200

plt.figure(figsize=(10,5))

# original
plt.subplot(1,2,1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis("off")

# mask
plt.subplot(1,2,2)
plt.imshow(mask, cmap="gray")
plt.title("Bright Pixels Mask")
plt.axis("off")

plt.show()
```

```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])

mask = arr > 25

print("mask: ", mask)
print("arr[mask]: ", arr[mask])

#notlar
#mask boyutu array ile aynı olmalıdır.
```

Mantıksal filtreleme

```python
import numpy as np

arr = np.array([5, 12, 18, 25, 30, 45])

mask = (arr > 10) & (arr < 30)
print("mask: ",  mask)
print("arr[mask]: ",  arr[mask])
```

```python
import numpy as np

pixels = np.array([40, 90, 220, 240, 130, 255, 70])

bright_mask = pixels > 200

print("Bright pixels:", pixels[bright_mask])
print("Normal pixels:", pixels[~bright_mask])
```
