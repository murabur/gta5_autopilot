dtype (data type), NumPy array içindeki her elemanın bellekte hangi veri tipi ile tutulduğunu belirtir. NumPy’nin performansının büyük kısmı bu kavrama dayanır çünkü veri tipi:

- kaç byte kullanılacağını
- sayının aralığını
 - matematiksel işlemlerin nasıl davranacağını
belirler.

Computer vision tarafında dtype çok kritik bir konudur çünkü görüntüler aslında sayısal matrislerdir.

```python
import numpy as np

arr = np.array([1,2,3,4])
print(arr.dtype)
```

```python
arr = np.array([1,2,3,4], dtype=np.float32)

print(arr)
print(arr.dtype)
```

```python
import numpy as np

arr = np.array([250], dtype=np.uint8)

print(arr + 10) #taşma oluşur. 250 + 10 = 260 -> 260 % 255 = 5 #0. indeksten başladığı için 0 - 1 - 2 - 3 - 4 
```

Computer Vision örneği

uint8 veri tipinin aralığı 0-255 olduğu için bu aralığı aşan
işlemler integer overflow üretir. NumPy bu durumda değeri
mod 256 alarak wrap-around davranışı gösterir.

```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
import os

current_dir = os.getcwd()

parent_dir = os.path.dirname(current_dir)
parent_dir = os.path.dirname(parent_dir)

img_path = os.path.join(parent_dir, "assets", "ornek_goruntu.png")

print(f"Erişilen Yol: {img_path}")

img = cv2.imread(img_path)

if img is None:
    print("Görüntü okunamadı!")
    exit()

print("Görüntü başarıyla yakalandı")

bright = img.copy() +  30

plt.figure(figsize=(10,5))

# original
plt.subplot(1,2,1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis("off")

# mask
plt.subplot(1,2,2)
plt.imshow(cv2.cvtColor(bright, cv2.COLOR_BGR2RGB))
plt.title("Overflow(wrap - around)")
plt.axis("off")

plt.show()

print("dtype:", img.dtype)
print("min:", img.min())
print("max:", img.max())

print("overflow min:", bright.min())
print("overflow max:", bright.max())
```
