# ndarray
Numpy temel veri yapısıdır. N-dimensional array(çok boyutlu dizi) anlamına gelir.

- Tüm elemanlar aynı veri tipindedir.
- Bellekte **contiguous (düz)** tutulur
- hızlı matematik işlemleri yapılabilir.

```python
import numpy as np

#np.array() array oluşturma yollarından biridir.


array_1d = np.array([1,2,3,4,5])
array_2d = np.array([[1,2,3],[1,2,3]])
array_3d = np.array([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])

print("array_1d:", array_1d, "\n")
print("array_2d: \n", array_2d, "\n")
print("array_3d:\n", array_3d, "\n")

```
