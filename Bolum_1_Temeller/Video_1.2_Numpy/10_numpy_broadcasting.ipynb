# Broadcasting

## 1. Broadcasting nedir

Broadcasting, NumPy'nin **farklı shape'lerdeki array'leri** belirli kurallarla **uyumlu hale getirip element-wise işlem yapabilmesidir**.

Kısa tanım:

```text
küçük shape
→ büyük shape'e uyarlanır
→ element-wise işlem yapılır

- Numpy çoğu durumda veriyi gerçekten kopyalamaz.
- Shapeleri mantıksal olarak uyumlu kabul eder
- Sonra işlemi uygular

2. Broadcasting neden gerekli

Broadcasting olmasaydı, çok basit işlemler için bile aynı boyutta array üretilmek zorunda kalınırdı.

````python
import numpy as np

arr = np.array([10, 20, 30, 40])

print("arr:", arr)
print("arr + 5:", arr + 5)
````

Numpy şu şekilde işlem yapıyormuş gibi davranır.

[10, 20, 30, 40] + [5, 5, 5, 5]

Aslında bu array'i oluşturmadan işlem yapar.



3. Broadcasting kuralları

NumPy iki array arasında işlem yaparken shape'leri sağdan sola karşılaştırır.

İki boyut karşılaştırılırken şu kurallar geçerlidir:

Kural 1

Boyutlar eşitse uyumludur.

(2,3)
(2,3)

Kural 2

Boyutlardan biri 1 ise uyumludur.

(2,3)
(1,3)

veya

(4,5,3)
(1,1,3)

Kural 3

Boyutlar eşit değilse ve hiçbirisi 1 değilse broadcasting yapılamaz.

(2,3)
(2,4)

uyumsuzdur.

4. Broadcasting nasıl düşünülmeli

- Adım 1
- Shape'leri yazdır.

#

- Adım 2
- Shape'leri sağdan sola hizala.
#
- Adım 3
- Her eksen için şunu kontrol et:
- eşit mi?
- biri 1 mi?

Eğer her eksende bu koşullardan biri sağlanıyorsa işlem yapılabilir.

Basit örnek: Skaler/Scalar Broadcasting

````python
import numpy as np

arr = np.array([1, 2, 3, 4])

print("arr:", arr)
print("arr + 10:", arr + 10)
print("arr * 2:", arr * 2)
````

````python
import numpy as np

a = np.array([10, 20, 30])
b = np.array([1, 2, 3])

print("a:", a)
print("b:", b)
print("a + b:", a + b)
````

````python
import numpy as np

A = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

b = np.array([10, 20, 30])

print("A:\n", A)
print("b:", b)
print("A + b:\n", A + b)

print(f"A.shape {A.shape}") #(2,3)
print(f"b.shape {b.shape}") #(3,)


#NumPy bunu sağdan sola şöyle hizalar:
#(2,3)
#(1,3)
````

````python
import numpy as np

A = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

b = np.array([10, 20])

print("A.shape:", A.shape)
print("b.shape:", b.shape)

print(A + b)
````

Sebep:

(2,3)
(1,2)

Sağdan sola karşılaştırınca son boyut:

3 vs 2

Ne eşit, ne de biri 1.

Bu yüzden broadcasting yapılamaz.

````python
import numpy as np

A = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

v = np.array([10, 20])

print("A:\n", A)
print("v:", v)
````

````python
import numpy as np

A = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

v = np.array([10, 20])

print("v[:, None]:\n", v[:, None])
print("v[:, None].shape:", v[:, None].shape)
print("A + v[:, None]:\n", A + v[:, None])
````

````python
import numpy as np

v = np.array([10, 20])

print(f"v.shape {v.shape}")
print(f"v[:, None] : \n {v[:, None]} \n")
print(f"v[:, None].shape : {v[:, None].shape} \n")
print(f"v[None, :] : {v[None, :]} \n")
print(f"v[None, :].shape: {v[None, :].shape}\n ")
````

````python
import numpy as np

v = np.array([1, 2, 3])

print("v:", v)
print("v.shape:", v.shape)

print("v[:, None]:\n", v[:, None])
print("v[:, None].shape:", v[:, None].shape)

print("v[None, :]:\n", v[None, :])
print("v[None, :].shape:", v[None, :].shape)
````

````python
v_col = v[:, np.newaxis]  #np.newaxis ile none aynıdır

print(v_col)
print(v_col.shape)
````
