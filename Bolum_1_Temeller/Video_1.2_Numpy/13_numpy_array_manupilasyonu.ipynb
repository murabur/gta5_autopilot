```python
#reshape, array'in shape'ini değiştirir. Veri aynı kalır, sadece düzen değişir.

import numpy as np

arr = np.array([1,2,3,4,5,6])

print("arr:", arr)
print("arr.shape:", arr.shape)

reshaped = arr.reshape(2,3)

print("reshaped:\n", reshaped)
print("reshaped.shape:", reshaped.shape)

#arr.reshape(4,2) olmaz. Toplam eleman sayısı değişmez.
```

```python
import numpy as np

arr = np.array([1,2,3,4,5,6])

print(arr.reshape(2, -1))
print(arr.reshape(-1, 3))

#-1 kullanmak boyutu sen hesapla demek
```

```python
#flatten array'i tek boyuta indirir ANCAK copy üretir.
import numpy as np

A = np.array([
    [1,2,3],
    [4,5,6]
])

flat = A.flatten()
flat[0] = 999

print("A:\n", A)
print("flat:", flat)
```

```python
#ravel flatten ile aynı işi yapar ama mümkünse view döndürür.

import numpy as np

A = np.array([
    [1,2,3],
    [4,5,6]
])

rav = A.ravel()

print("A:\n", A)
print("rav:", rav)
```

```python
#transpose(), eksenlerin yerini değiştirir.
#2d için en bilinen anlamı satırları sütün yapar, sütunları satır yapar.

import numpy as np

A = np.array([
    [1,2,3],
    [4,5,6]
])

print("A:\n", A)
print("A.shape:", A.shape)

T = A.transpose()

#A.T şeklinde kullanımı da vardır.

print("T:\n", T)
print("T.shape:", T.shape)
```

```python
#np.newaxis / None yeni eksen eklerler

#sütun vektörü yapımı
import numpy as np

v = np.array([1,2,3])

print("v:", v)
print("v.shape:", v.shape)

v_col = v[:, np.newaxis]

print("v_col:\n", v_col)
print("v_col.shape:", v_col.shape)
```

```python
#satır vektörü yapımı

import numpy as np

v = np.array([1,2,3])

v_row = v[np.newaxis, :]

print("v_row:\n", v_row)
print("v_row.shape:", v_row.shape)
```

```python
#np.expand_dims() bu da yeni eksen ekler.

import numpy as np

v = np.array([1,2,3])

col = np.expand_dims(v, axis=1)
row = np.expand_dims(v, axis=0)

print("col:\n", col)
print("col.shape:", col.shape)

print("row:\n", row)
print("row.shape:", row.shape)
```

```python
#squeeze, boyutu 1 olan eksenleri kaldırır.

import numpy as np

A = np.array([[[1,2,3]]])

print("A.shape:", A.shape)

B = np.squeeze(A)

print("B:", B)
print("B.shape:", B.shape)
```

```python
#concatenate() mevcut eksen boyunca arrayleri birleştirir.
#concatenate için diğer eksenler aynı olmalı

#axis = 0 durumu 

import numpy as np

A = np.array([
    [1,2],
    [3,4]
])

B = np.array([
    [5,6],
    [7,8]
])

C = np.concatenate([A,B], axis=0)

print("C:\n", C)
print("C.shape:", C.shape)
```

```python
#axis = 1 durumu

import numpy as np

A = np.array([
    [1,2],
    [3,4]
])

B = np.array([
    [5,6],
    [7,8]
])

C = np.concatenate([A,B], axis=1)

print("C:\n", C)
print("C.shape:", C.shape)
```

```python
#stack() arrayleri yeni bir eksen oluşturarak birleştirir.

import numpy as np

a = np.array([1,2,3])
b = np.array([4,5,6])

s = np.stack([a,b], axis=0)

print("s:\n", s)
print("s.shape:", s.shape)
```

```python
#axis = 1 

import numpy as np

a = np.array([1,2,3])
b = np.array([4,5,6])

s = np.stack([a,b], axis=1)

print("s:\n", s)
print("s.shape:", s.shape)
```

```python
#hstack ve vstack yatat ve dikey birleştirme için kısayoldur.

import numpy as np

A = np.array([[1,2],[3,4]])
B = np.array([[5,6],[7,8]])

print(np.hstack([A,B]))

```

```python
import numpy as np

A = np.array([[1,2],[3,4]])
B = np.array([[5,6],[7,8]])

print(np.vstack([A,B]))
```

```python
#split parçalara böler
import numpy as np

arr = np.array([1,2,3,4,5,6])

parts = np.split(arr, 3)

print(parts)
```

```python
import numpy as np

A = np.array([
    [1,2,3,4],
    [5,6,7,8]
])

parts = np.split(A, 2, axis=1)

for p in parts:
    print(p)
```

```python
#hsplit ve vsplit kısayol split fonksiyonlarıdır.

import numpy as np

A = np.array([
    [1,2,3,4],
    [5,6,7,8]
])

parts = np.hsplit(A, 2)

for p in parts:
    print(p)
```

```python
import numpy as np

A = np.array([
    [1,2],
    [3,4],
    [5,6],
    [7,8]
])

parts = np.vsplit(A, 2)

for p in parts:
    print(p)
```
