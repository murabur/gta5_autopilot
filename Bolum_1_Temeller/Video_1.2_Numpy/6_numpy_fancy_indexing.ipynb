Fancy Indexing, index olarak liste veya array kullanmaktır. Normal indexingden farkı budur. Normal indexing ve slice genelde view döndürür. Fancy Indexing her zaman copy üretir.
- normal indexing → tek index
- slicing → aralık
- fancy indexing → istediğin index listesi

```python
import numpy as np

arr = np.array([10,20,30,40,50])

arr[[0,2,4]] #belirli elemanları seçmek
```

Fancy indexing her zaman için copy üretir.

```python
import numpy as np


arr = np.array([
    [10,20,30],
    [40,50,60],
    [70,80,90]
])

arr[[0,2]] #belirli satırlar
```

```python
import numpy as np


arr = np.array([
    [10,20,30],
    [40,50,60],
    [70,80,90]
])

arr[:, [0,2]] #belirli sütunlar
```

```python
import numpy as np


arr = np.array([
    [10,20,30],
    [40,50,60],
    [70,80,90]
])

arr[[0,2], [1,2]] #satır ve sütun seçimi
```

Koordinat bazlı seçim

```python
import numpy as np


arr = np.array([
    [10,20,30],
    [40,50,60],
    [70,80,90]
])

rows = np.array([0,2])
cols = np.array([1,2])

arr[rows, cols]
```
