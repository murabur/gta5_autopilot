```python
arr = np.array([1,2,3])

#verilen np.array ile aynı boyutta 0 veya 1lerden oluşan np.array oluşturur
arr_zeros_like = np.zeros_like(arr)
arr_ones_like = np.ones_like(arr)

print(arr_zeros_like, "\n")
print(arr_ones_like, "\n")
```
