En temek istatistik fonksiyonları

- np.sum
- np.mean
- np.min
- np.max
- np.std
- np.var

```python
#np.sum toplama

import numpy as np

arr = np.array([10, 20, 30, 40])

print("arr:", arr)
print("np.sum(arr):", np.sum(arr))
```

```python
A = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print("A:\n", A)
print("np.sum(A):", np.sum(A))
```

```python
#Axis ile
A = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print("np.sum(A, axis=0):", np.sum(A, axis=0))
print("np.sum(A, axis=1):", np.sum(A, axis=1))
```

```python
import numpy as np
arr = np.array([10, 20, 30, 40])

print("arr:", arr)
print("np.mean(arr):", np.mean(arr))
```

```python
#varyans = veriler birbirinden ne kadar dağınık
import numpy as np 

arr = np.array([1, 2, 3, 4, 5])

print("arr:", arr)
print("np.var(arr):", np.var(arr))
```
